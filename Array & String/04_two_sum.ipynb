This is **LeetCode Problem 1: Two Sum**. It is the most famous coding interview problem and serves as the classic introduction to using **Hash Maps** for optimization.

Here is the full breakdown of the problem, logic, Brute Force vs. Optimal approaches, and the Java code.

---

### 1. Problem Statement
Given an array of integers `nums` and an integer `target`, return *indices* of the two numbers such that they add up to `target`.

*   You may assume that each input would have **exactly one solution**.
*   You may not use the same element twice.

**Example:**
*   **Input:** `nums = [2, 7, 11, 15]`, `target = 9`
*   **Output:** `[0, 1]`
*   **Explanation:** `nums[0] + nums[1] == 9`, so we return `[0, 1]`.

---

### 2. Approaches

#### Approach 1: Brute Force (Naive)
**Logic:**
Check every possible pair of numbers to see if they add up to the target.
1.  Pick the first number `nums[i]`.
2.  Loop through the rest of the array (`j = i + 1`) to find a second number.
3.  If `nums[i] + nums[j] == target`, return indices `[i, j]`.

**Code:**
```java
public int[] twoSumBruteForce(int[] nums, int target) {
    for (int i = 0; i < nums.length; i++) {
        // Start j at i + 1 so we don't reuse the same element
        for (int j = i + 1; j < nums.length; j++) {
            if (nums[i] + nums[j] == target) {
                return new int[] { i, j };
            }
        }
    }
    return new int[] {}; // Should not happen based on problem description
}
```
*   **Time Complexity:** $O(n^2)$ — Nested loops.
*   **Space Complexity:** $O(1)$ — No extra data structures used.

---

#### Approach 2: HashMap (Optimal / One-Pass)
**Logic:**
Instead of scanning the whole array to find the second number, we can use a **HashMap** to remember the numbers we have already seen and their indices.

The math equation is: `current_number + x = target`.
Therefore: `x = target - current_number`.
We call `x` the **complement**.

**Algorithm:**
1.  Create an empty HashMap `map` to store `{value : index}`.
2.  Iterate through the array (`i` from `0` to `n`).
3.  Calculate `complement = target - nums[i]`.
4.  **Check:** Does the HashMap already contain the `complement`?
    *   **Yes:** We found the pair! Return the index from the map and current index `i`.
    *   **No:** Store the current number and its index `map.put(nums[i], i)` and continue.

**Why this works:**
We only pass through the array once. As we move forward, we look *backward* into the map to see if the "partner" for the current number is already there.

---

### 3. Java Solution (Optimal)

```java
import java.util.HashMap;
import java.util.Map;

class Solution {
    public int[] twoSum(int[] nums, int target) {
        // Map to store: Key = Number, Value = Index
        Map<Integer, Integer> numMap = new HashMap<>();

        for (int i = 0; i < nums.length; i++) {
            int currentNum = nums[i];
            int complement = target - currentNum;

            // Check if the complement exists in the map
            if (numMap.containsKey(complement)) {
                // If found, return the index of the complement and current index
                return new int[] { numMap.get(complement), i };
            }

            // Otherwise, add the current number and index to the map
            numMap.put(currentNum, i);
        }

        // Return empty if no solution found (problem guarantees a solution though)
        return new int[] {}; 
    }
}
```

---

### 4. Dry Run (Visualizing the Logic)

**Input:** `nums = [2, 11, 7, 15]`, `target = 9`

1.  **Index 0 (`val = 2`):**
    *   `complement` = 9 - 2 = **7**.
    *   Is 7 in Map? **No**.
    *   Add to Map: `{2 : 0}`.
2.  **Index 1 (`val = 11`):**
    *   `complement` = 9 - 11 = **-2**.
    *   Is -2 in Map? **No**.
    *   Add to Map: `{2 : 0, 11 : 1}`.
3.  **Index 2 (`val = 7`):**
    *   `complement` = 9 - 7 = **2**.
    *   Is 2 in Map? **Yes!** (It's at index 0).
    *   **Return:** `[0, 2]`.

### 5. Complexity Analysis

| Metric | Complexity | Explanation |
| :--- | :--- | :--- |
| **Time Complexity** | **$O(n)$** | We traverse the list containing $n$ elements exactly once. Hash Map lookups take $O(1)$ time on average. |
| **Space Complexity** | **$O(n)$** | In the worst case, we might store almost every element in the HashMap before finding the match. |

### Summary
*   Use **Brute Force** if $n$ is extremely small (e.g., < 100) or if you cannot use extra memory.
*   Use **HashMap** for almost all real-world scenarios as it is significantly faster.
