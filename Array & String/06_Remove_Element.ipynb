## Remove Element
### 1. Problem Statement

Given an integer array `nums` and an integer `val`, remove all occurrences of `val` in `nums` **in-place**. The order of the elements may be changed. Then return the number of elements in `nums` which are not equal to `val`.

Consider the number of elements in `nums` which are not equal to `val` be `k`. To get accepted, you need to do the following:
1. Change the array `nums` such that the first `k` elements of `nums` contain the elements which are not equal to `val`.
2. The remaining elements of `nums` are not important as well as the size of `nums`.
3. Return `k`.

**Example 1:**
*   **Input:** `nums = [3,2,2,3], val = 3`
*   **Output:** `2, nums = [2,2,_,_]`

**Example 2:**
*   **Input:** `nums = [0,1,2,2,3,0,4,2], val = 2`
*   **Output:** `5, nums = [0,1,4,0,3,_,_,_]`

---

### 2. Logic (Two Pointer Approach)

Since we must solve this **in-place** (without creating a new array) and with $O(1)$ extra memory, we use the **Two Pointer** technique.

**The Concept:**
We need one pointer to iterate through the array (`i`) and another pointer (`k`) to keep track of the position where the *next valid number* should be written.

1.  Initialize `k = 0`. This represents the number of valid elements found so far.
2.  Iterate through the array with `i` from `0` to `nums.length - 1`.
3.  Check the value at `nums[i]`:
    *   If `nums[i]` is equal to `val`: It is "trash". We ignore it and just move `i` forward.
    *   If `nums[i]` is **not** equal to `val`: It is a "keeper". We write it to position `k` (`nums[k] = nums[i]`) and then increment `k`.
4.  Return `k`.

**Visualizing `nums = [3, 2, 2, 3], val = 3`**

*   `i=0`, `nums[0]=3`. Equals val. **Skip**. (`k` stays 0).
*   `i=1`, `nums[1]=2`. Not val. **Write** `nums[k] = 2`. Increment `k` to 1. (Array is now `[2, 2, 2, 3]`).
*   `i=2`, `nums[2]=2`. Not val. **Write** `nums[k] = 2`. Increment `k` to 2. (Array is now `[2, 2, 2, 3]`).
*   `i=3`, `nums[3]=3`. Equals val. **Skip**.
*   **Result:** Return `k = 2`. The first 2 elements are `[2, 2]`.

---

### 3. Java Solution

```java
class Solution {
    public int removeElement(int[] nums, int val) {
        // k will track the position of the next non-val element
        int k = 0;

        for (int i = 0; i < nums.length; i++) {
            // If the current number is NOT the value we want to remove
            if (nums[i] != val) {
                // Move it to the front of the array at index k
                nums[k] = nums[i];
                // Increment k to point to the next open spot
                k++;
            }
            // If nums[i] == val, we do nothing and just let the loop continue
        }

        // k is now the count of valid elements
        return k;
    }
}
```

---

### 4. Complexity Analysis

| Metric | Complexity | Explanation |
| :--- | :--- | :--- |
| **Time Complexity** | **$O(N)$** | We traverse the array exactly once (single pass). |
| **Space Complexity** | **$O(1)$** | We modify the array in-place and don't use any extra data structures. |

### Why is this efficient?
We don't actually "delete" the items (which would require shifting the whole array left every time, taking $O(N^2)$). Instead, we just **overwrite** the values at the beginning of the array with the values we want to keep. The "trash" values are left at the end of the array, but since the problem says "remaining elements are not important," we ignore them.
