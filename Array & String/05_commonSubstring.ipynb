
# 14. Longest Common Prefix

##  Problem Statement

Write a function to find the **longest common prefix string** amongst an array of strings.

If there is **no common prefix**, return an empty string `""`.

---

##  Input

* An array of strings `strs`

---

##  Output

* A string representing the **longest common prefix**

---

##  Examples

### Example 1

```
Input: strs = ["flower","flow","flight"]
Output: "fl"
```

### Example 2

```
Input: strs = ["dog","racecar","car"]
Output: ""
```

**Explanation:**
There is no common prefix among the input strings.

---

##  Constraints

* `1 <= strs.length <= 200`
* `0 <= strs[i].length <= 200`
* `strs[i]` consists of only lowercase English letters if it is non-empty



This is **LeetCode 14: Longest Common Prefix**.

Here are the two best ways to solve this: the **Horizontal Scanning** method (standard) and the **Sorting** method (clever trick).

---

### Approach 1: Sorting (The "Smart" Approach)
**Logic:**
If you sort an array of strings alphabetically, the strings with the most differences will end up at the **beginning** and the **end** of the sorted list.
*   Therefore, the longest common prefix must be the common prefix between the **First** and the **Last** string.
*   We don't need to check the strings in the middle.

**Example:**
Input: `["flower", "flow", "flight"]`
1.  **Sort:** `["flight", "flow", "flower"]`
2.  **Compare:** First (`"flight"`) and Last (`"flower"`).
3.  **Result:** "fl"

**Java Code:**
```java
import java.util.Arrays;

class Solution {
    public String longestCommonPrefix(String[] strs) {
        // 1. Sort the array
        Arrays.sort(strs);
        
        // 2. Grab the first and last strings
        String s1 = strs[0];
        String s2 = strs[strs.length - 1];
        
        int idx = 0;
        
        // 3. Compare characters while they match
        while(idx < s1.length() && idx < s2.length()){
            if(s1.charAt(idx) == s2.charAt(idx)){
                idx++;
            } else {
                break;
            }
        }
        
        // 4. Return the substring from 0 to the matching index
        return s1.substring(0, idx);
    }
}
```
*   **Time Complexity:** $O(N \cdot L \cdot \log N)$ (Sorting takes the most time).
*   **Space Complexity:** $O(1)$ (Assuming we don't count the result string).

---

### Approach 2: Horizontal Scanning (Standard / Linear)
**Logic:**
1.  Take the first string (`strs[0]`) and assume it is the **prefix**.
2.  Compare it with the second string. If the second string doesn't start with the prefix, **shorten the prefix** by one character from the end.
3.  Repeat this until the second string starts with the prefix.
4.  Move to the third string and repeat.

**Example:** `["flower", "flow", "flight"]`
1.  Prefix = "flower"
2.  Compare with "flow":
    *   Does "flow" start with "flower"? No. Shorten -> "flowe" -> "flow".
    *   Does "flow" start with "flow"? Yes.
    *   New Prefix = "flow"
3.  Compare with "flight":
    *   Does "flight" start with "flow"? No. Shorten -> "flo" -> "fl".
    *   Does "flight" start with "fl"? Yes.
4.  Result = "fl"

**Java Code:**
```java
class Solution {
    public String longestCommonPrefix(String[] strs) {
        if (strs.length == 0) return "";

        // Start by assuming the whole first string is the prefix
        String prefix = strs[0];

        // Loop through the rest of the strings
        for (int i = 1; i < strs.length; i++) {
            
            // While the current string (strs[i]) does not start with prefix
            // indexOf returns 0 if it is a prefix, -1 or >0 otherwise
            while (strs[i].indexOf(prefix) != 0) {
                // Shorten the prefix by removing the last character
                prefix = prefix.substring(0, prefix.length() - 1);

                // If prefix becomes empty, there is no common prefix
                if (prefix.isEmpty()) return "";
            }
        }

        return prefix;
    }
}
```

*   **Time Complexity:** $O(S)$, where $S$ is the sum of all characters in all strings. In the worst case, we compare every character.
*   **Space Complexity:** $O(1)$ (We modify the prefix variable in place).

---

### Summary
*   **Sorting Approach:** Easier to write in an interview, very clean logic. Slightly slower due to sorting overhead ($N \log N$), but negligible for $N=200$.
*   **Horizontal Scanning:** The theoretically faster approach ($O(N)$) for large arrays because it avoids sorting.

**Recommendation:**
Use **Horizontal Scanning** (Approach 2) if you want to show off understanding of string manipulation and `indexOf`. Use **Sorting** (Approach 1) if you want the code to be short and easy to remember.
